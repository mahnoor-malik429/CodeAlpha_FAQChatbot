```python
# Import Libraries
import nltk
import string
import gradio as gr

from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# Download NLTK Data
nltk.download('punkt')
nltk.download('punkt_tab')



# FAQ Questions and Answers
faq_data = {

    "what is ai":
    "Artificial Intelligence is a technology that enables machines to simulate human intelligence.",

    "what is machine learning":
    "Machine Learning is a branch of AI that allows systems to learn from data.",

    "what is deep learning":
    "Deep Learning is a subset of machine learning based on neural networks.",

    "what is python":
    "Python is used for web development, artificial intelligence, automation, and data science.",

    "what is nlp":
    "Natural Language Processing helps computers understand and process human language.",

    "what is chatbot":
    "A chatbot is an AI application that interacts with users through conversation.",

    "what is data science":
    "Data Science is the process of analyzing and extracting insights from data.",

    "what is computer vision":
    "Computer Vision enables machines to understand and analyze images and videos.",

    "what is neural network":
    "A neural network is a deep learning model inspired by the structure of the human brain."
}



# Text Preprocessing Function
def preprocess_text(text):

    text = text.lower()

    tokens = word_tokenize(text)

    cleaned_tokens = []

    for word in tokens:

        if word not in string.punctuation:
            cleaned_tokens.append(word)

    return " ".join(cleaned_tokens)



# Prepare Questions
questions = list(faq_data.keys())

processed_questions = [preprocess_text(q) for q in questions]



# TF-IDF Vectorization
vectorizer = TfidfVectorizer()

question_vectors = vectorizer.fit_transform(processed_questions)



# Chatbot Function
def chatbot_response(user_question):

    user_question = preprocess_text(user_question)

    user_vector = vectorizer.transform([user_question])

    similarity = cosine_similarity(user_vector, question_vectors)

    best_match_index = similarity.argmax()

    similarity_score = similarity[0][best_match_index]

    # Low Similarity Handling
    if similarity_score < 0.2:
        return "Sorry, I could not understand your question. Please ask another AI-related question."

    best_question = questions[best_match_index]

    answer = faq_data[best_question]

    return answer


custom_css = """

body {
    background: linear-gradient(to right, #0f172a, #1e293b);
}

.gradio-container {
    max-width: 950px !important;
    margin: auto;
    padding-top: 20px;
    font-family: 'Segoe UI', sans-serif;
}

.main-title {
    text-align: center;
    font-size: 42px;
    font-weight: bold;
    color: white;
    margin-bottom: 10px;
}

.sub-title {
    text-align: center;
    font-size: 18px;
    color: #d1d5db;
    margin-bottom: 30px;
}

.chat-box {
    background: white;
    padding: 30px;
    border-radius: 20px;
    box-shadow: 0px 6px 25px rgba(0,0,0,0.25);
}

textarea {
    border-radius: 12px !important;
}

button {
    background: #2563eb !important;
    color: white !important;
    border-radius: 10px !important;
    height: 52px;
    font-size: 17px !important;
    font-weight: 600 !important;
    border: none !important;
}

button:hover {
    background: #1d4ed8 !important;
}

footer {
    visibility: hidden;
}
"""



# Build Interface
with gr.Blocks(css=custom_css) as demo:

    gr.HTML("""
    <div class="main-title">
        AI FAQ Chatbot
    </div>

    <div class="sub-title">
        An intelligent chatbot that answers frequently asked questions using NLP and cosine similarity techniques.
    </div>
    """)

    with gr.Column(elem_classes="chat-box"):

        user_input = gr.Textbox(
            lines=4,
            placeholder="Ask your AI-related question here...",
            label="User Question"
        )

        chatbot_output = gr.Textbox(
            lines=5,
            label="Chatbot Response",
            show_copy_button=True
        )

        ask_button = gr.Button("Get Answer")

        ask_button.click(
            fn=chatbot_response,
            inputs=user_input,
            outputs=chatbot_output
        )

# Launch Application
demo.launch()
```
